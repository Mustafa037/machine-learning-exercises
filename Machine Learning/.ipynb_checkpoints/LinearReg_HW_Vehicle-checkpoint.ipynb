```python
# kategorik sütunları seç
categorical_cols = ["fuel", "transmission"]

# one-hot encoding
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

df_encoded.head()
```

```python
df_encoded["age"] = df_encoded["year"]
```

```python
df_encoded["age"] = df["age"]
df_encoded["age"]
```

```python
df_encoded.corr(numeric_only=True)
```

```python
corr_with_price = df_encoded.corr(numeric_only=True)["selling_price"].sort_values(ascending=False)
corr_with_price.head(15)   # pozitif en yüksek 15
```
